##### Elective Surgery Dataset Overview

```python
print("="*50)
print("Elective Surgery Dataset Overview")
print("="*50)

print("\nFirst 5 Rows:")
print(es.head())

print(f"Rows: {es.shape[0]}")
print(f"Columns: {es.shape[1]}")
print(f"Shape: {es.shape}")

print("\nColumn Names:")
print(es.columns.tolist())

print("\nData Types:")
print(es.dtypes)

print("\nMissing Values:")
print(es.isnull().sum())

print("\nDuplicate Rows:")
print(es.duplicated().sum())

print("\nUnique Values:")
print(es.nunique())

for col in es.columns:
    print(f"\n{col}")
    print(es[col].unique())

```

```python
# Check for missing values in the expected_los_days column
es[es["expected_los_days"].isna()]
```

```python
# Calculate the median for each procedure
procedure_median = es.groupby("procedure_group")["expected_los_days"].median()

# Fill missing values
es["expected_los_days"] = es["expected_los_days"].fillna(es["procedure_group"].map(procedure_median))

# check for missing values in the expected_los_days column after filling
es[es["expected_los_days"].isna()]
```

```python
# Check for missing values in the cancellation_reason column
es[es["cancellation_reason"].isna()]
```

```python
# Fill missing values in the time_to_admission column with 0
es["cancellation_reason"] = es["cancellation_reason"].fillna("Completed Surgery")
es["cancellation_reason"].isna().sum()
```

```python
# Convert the date column to datetime format
es["surgery_date"] = pd.to_datetime(es["surgery_date"])
```

##### Feature Engineering

```python
sr["staffing_year"] = sr["date"].dt.year
sr["staffing_month"] = sr["date"].dt.month_name()
sr["staffing_quarter"] = sr["date"].dt.quarter
sr["staffing_weekday"] = sr["date"].dt.day_name()
```

```python
es["surgery_year"] = es["surgery_date"].dt.year
es["surgery_month"] = es["surgery_date"].dt.month_name()
es["surgery_quarter"] = es["surgery_date"].dt.quarter
es["surgery_weekday"] = es["surgery_date"].dt.day_name()
```

```python
ad["admission_year"] = ad["admission_datetime"].dt.year
ad["admission_month"] = ad["admission_datetime"].dt.month_name()
ad["admission_quarter"] = ad["admission_datetime"].dt.quarter
ad["admission_weekday"] = ad["admission_datetime"].dt.day_name()
ad["admission_hour"] = ad["admission_datetime"].dt.hour
```

```python
ad["discharge_year"] = ad["discharge_datetime"].dt.year
ad["discharge_month"] = ad["discharge_datetime"].dt.month_name()
ad["discharge_quarter"] = ad["discharge_datetime"].dt.quarter
ad["discharge_weekday"] = ad["discharge_datetime"].dt.day_name()
ad["discharge_hour"] = ad["discharge_datetime"].dt.hour
```

```python
bd["bedding_year"] = bd["datetime"].dt.year
bd["bedding_month"] = bd["datetime"].dt.month_name()
bd["bedding_quarter"] = bd["datetime"].dt.quarter
bd["bedding_weekday"] = bd["datetime"].dt.day_name()
bd["bedding_hour"] = bd["datetime"].dt.hour
```

```python
oa["arrival_year"] = oa["arrival_datetime"].dt.year
oa["arrival_month"] = oa["arrival_datetime"].dt.month_name()
oa["arrival_quarter"] = oa["arrival_datetime"].dt.quarter
oa["arrival_weekday"] = oa["arrival_datetime"].dt.day_name()
oa["arrival_hour"] = oa["arrival_datetime"].dt.hour
```

##### Normalisation Across Sources

```python
set(ad["hospital_id"]) == set(hr["hospital_id"])
set(bd["hospital_id"]) == set(hr["hospital_id"])
set(oa["hospital_id"]) == set(hr["hospital_id"])
set(es["hospital_id"]) == set(hr["hospital_id"])
set(sr["hospital_id"]) == set(hr["hospital_id"])
```

```python
ad["admission_datetime"] = ad["admission_datetime"].dt.normalize()
ad["discharge_datetime"] = ad["discharge_datetime"].dt.normalize()
bd["datetime"] = bd["datetime"].dt.normalize()
oa["arrival_datetime"] = oa["arrival_datetime"].dt.normalize()
es["surgery_date"] = es["surgery_date"].dt.normalize()
```
